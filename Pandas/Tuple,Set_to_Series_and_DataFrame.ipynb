Set to Series and **Dataframe**

```python
import pandas as pd
s1={1,2,3,4,5,7,10}
print(s1)
print(type(s1))
```

```python
s=pd.Series(list(s1))
print(s)
```

```python
df=pd.DataFrame(s1,columns=['set value'])
print(df)
```

# **Tuple to Series and datFrame**

```python
tuple=(1,2,3,4,5,6,True,False)
s=pd.Series(tuple)
print(s)
```

```python
df=pd.DataFrame(tuple,columns=['Tuple Value'])
print(df)
```
