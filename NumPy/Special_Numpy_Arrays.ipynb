# **Zeros**

```python
import numpy as np
ar_zero = np.zeros(4)
ar_zero1 = np.zeros((3,4))
print(ar_zero)
print()
print(ar_zero1)
```

# Ones

```python
ar_one=np.ones(4)
print(ar_one)
ar_one1=np.ones((3,4))
print()
print(ar_one1)
```

# **Empty**

```python
ar_empty=np.empty(4)
print(ar_empty)
```

# **Range**

```python
ar_range=np.arange(4)
print(ar_range)
```

# **Diagonal**

```python
ar_diagonal=np.eye(3)
print(ar_diagonal)
```

```python
ar_dia1=np.eye(3,5)
print(ar_dia1)
```

# **Linespace**

```python
ar_line=np.linspace(1,10,num=5)
print(ar_line)
```
