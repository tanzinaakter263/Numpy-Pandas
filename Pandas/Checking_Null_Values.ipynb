```python
import numpy as np
import pandas as pd
```

```python
data=np.random.rand(10,3)
for col in range(data.shape[1]):
  null_indices=np.random.choice(10,3,replace=False)
  data[null_indices,col]=np.nan
df=pd.DataFrame(data,columns=['Column_1','Column_2','Column_3'])
df.head()
```

```python
df.fillna(
    {
        'Column_1':1,
        'Column_2':2,
        'Column_3':3
    }
)
```

```python
df.dropna(how='any')   #Default any
```

```python
df.dropna(how='all')
```

```python
df.dropna(thresh=2)
```

```python
df.dropna(thresh=3)
```

```python
df.dropna(thresh=4)
```

# Filling Null with central tendency

## Mean(Average)= Sum of all Values/Number of values
## Median=middle values
## mode=most frequently in a dataset

```python
df.fillna(df.mean())
```

```python
df
```

```python
df.mean()
```

```python
df.fillna(df.median())   #for outlier
```
