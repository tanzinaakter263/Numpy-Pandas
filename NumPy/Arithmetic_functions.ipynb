```python
import numpy as np
var=np.array([3,2,1,4,5,3,2])
print("Minimum: ",np.min(var))
print("Maximum: ",np.max(var))
print("Square: ", np.sqrt(var))
```

```python
var=np.array([3,2,1,4,5,3,2])
print("Minimum: ",np.min(var),np.argmin(var))
print("Maximum: ",np.max(var),np.argmax(var))
```

```python
var1=np.array([[2,1,3],[9,5,6]])         #axis=0 col axis =1 row
print(np.min(var1,axis=0))
```

```python
var2=np.array([1,2,3])
print(np.sin(var2))
print(np.cos(var2))
print(np.cumsum(var2))
```
