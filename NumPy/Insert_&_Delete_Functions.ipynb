**Insert**

```python
import numpy as np
var=np.array([1,2,3,4])
print(var)
print(type(var))
insert=np.insert(var,2,40)
print(insert)
```

```python
var=np.array([1,2,3,4])
print(var)
print(type(var))
insert=np.insert(var,[2,4],40.6)
print(insert)
```

```python
var1=np.array([[1,2,3],[1,2,3]])
#print(var1)

insert1=np.insert(var1,2,[22,23],axis=1)
print(insert1)
```

```python
var=np.array([1,2,3,4])
print(var)

x=np.append(var,6.5)
print(x)
```

```python
var1=np.array([[1,2,3],[1,2,3]])
v1=np.append(var1,[[44,45,46]],axis=0)
print(v1)
```
