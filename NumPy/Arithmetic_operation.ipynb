# **Arithmetic Operation**

```python
import numpy as np
var = np.array([1,2,3,4])
varadd=var+3
print(varadd)
```

```python
var1 = np.array([1,2,3,4])
var2 = np.array([1,2,3,4])
varadd=var1+var2
print(varadd)
```

```python
var1 = np.array([1,2,3,4])
var2 = np.array([1,2,3,4])
varadd=np.add(var1,var2)
print(varadd)
```

```python
var = np.array([1,2,3,4])
varadd=np.add(var+3)
print(varadd)
```

```python
var = np.array([1,2,3,4])
varadd=var-3
print(varadd)
```

```python
var = np.array([1,2,3,4])
varadd=var*3
print(varadd)
```

```python
var = np.array([1,2,3,4])
varadd=var/3
print(varadd)
```

```python
var = np.array([1,2,3,4])
varadd=var%3
print(varadd)
```

```python
var = np.array([1,2,3,4])
varadd=np.reciprocal(var)
print(varadd)
```

# **2D Array**

```python
var21 = np.array([[1,2,3,4],[1,2,3,4]])
var22 = np.array([[1,2,3,4],[1,2,3,4]])

print(var21)
print()
print(var22)
print()

varadd2=var21+var22
print(varadd2)
```

```python
var21 = np.array([[1,2,3,4],[1,2,3,4]])
var22 = np.array([[1,2,3,4],[1,2,3,4]])

print(var21)
print()
print(var22)
print()

varadd2=var21*var22
print(varadd2)
```
