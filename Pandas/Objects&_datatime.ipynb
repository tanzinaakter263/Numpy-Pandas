```python
#Create a list of dates
dates=['2023-10-04','2024-11-15','2025-01-01']
#Craete a DataFrame with a datetime column
df1=pd.DataFrame({'Date':dates})
#Check the data types of the 'Date' columns
df1['Date'].dtype
```
