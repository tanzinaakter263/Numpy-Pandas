# **Matrix**

```python
import numpy as np
var=np.matrix([[1,2,3],[1,2,3]])
print(var)
print(type(var))
```

```python
var=np.array([[1,2,3],[1,2,3]])
print(var)
print(type(var))
```

```python
var1=np.matrix([[1,2,3],[1,2,3]])
var2=np.matrix([[1,2,3],[1,2,3]])
print(var1)
print()
print(var2)
print()

print(var1+var2)
```

```python
var1=np.matrix([[1,2,3],[1,2,3]])
var2=np.matrix([[1,2,3],[1,2,3]])
print(var1)
print()
print(var2)
print()

print(var1-var2)
```

```python
var1=np.matrix([[1,2],[1,2]])
var2=np.matrix([[1,2],[1,2]])
print(var1)
print()
print(var2)
print()

print(var1*var2)
print()
print(var1.dot(var2))
```

```python
var1=np.array([[1,2,3],[1,2,3]])

print(var1)
print()
print(var1*var1)
```

# **Transpose Matrix**

```python
var=np.matrix([[1,2,3],[4,5,6]])
print(var)
print()
print(np.transpose(var))
print()
print(var.T)
```

# **swapaxes**

```python
var=np.matrix([[1,2,3],[4,5,6]])
print(var)
print()
print(np.swapaxes(var,0,1))
```

```python
var2=np.matrix([[1,2],[3,4]])
print(var2)
print()
print(np.swapaxes(var2,0,1))
```

# **Inverse**

```python
var3=np.matrix([[1,2],[3,4]])
print(var3)
print()

print(np.linalg.inv(var3))
```

# **Power**

```python
var4=np.matrix([[1,2],[3,4]])
print(var4)
print()

print(np.linalg.matrix_power(var4,2))
print()

print(np.linalg.matrix_power(var4,0))
print()

print(np.linalg.matrix_power(var4,-2))
print()
```

# **determinate**

```python
var5=np.matrix([[1,2,3],[3,4,3],[1,2,3]])
print(var5)
print()
print(np.linalg.det(var5))
```
