# **Indexing**

```python
import numpy as np
var=np.array([9,8,7,6])
            # 0,1,2,3
            #-4,-3,-2,-1
print(var[1])
print(var[-3])
```

```python
var1=np.array([[9,8,7],[4,5,6]])
print(var1)
print(var1.ndim)
print()
print(var1[0,1])
```

```python
var2=np.array([[[1,2],[6,7]]])
print(var2)
print(var2.ndim)
print()
print(var2[0,1,1])
```

# Slicing

```python
import numpy as np
var=np.array([9,8,7,6,5,4,3])
         #    0,1,2,3,4,5,6
print(var)
print()
print("2 to 5: ",var[1:5])
print("2 to End: ",var[1:])
print("Start to 5: ",var[:5])
print("Stop: ",var[::])
print("Stop1: ",var[1:6:2])
```

```python
var1=np.array([[1,2,3,4,5],[9,8,7,6,5],[11,12,13,14,15]])
print(var1)
print()
print("8 to 5: ",var1[1,1:])
print("12 to 15: ", var1[2,1:])
```
