# **Rename & Concate DataFrame**

```python
df1=pd.DataFrame({'Age':[16,20],'Income':[4000,40000]})
df2=pd.DataFrame({'Age':[25,26],'Income':[8000,80000]})
```

```python
df1
```

```python
df2
```

```python
df1=df1.rename(columns={'Age':'AGE','Income':'INCOME'})
df1
```

```python
df2=df2.rename(columns={'Age':'AGE','Income':'INCOME'})
df2
```

```python
#concate
result = pd.concat([df1,df2],axis=0)   #row wise concate cause axis=0 is default
result
```

```python
#concate
result=pd.concat([df1,df2],axis=1)   #row wise concate cause axis=0 is default
result
```
