# **Data Types**

```python
import numpy as np
var=np.array([1,2,3,4,13,13,14,15])
print("Data types: ", var.dtype)
```

```python
var=np.array([1.8,2.9,3.5,4.0])
print("Data types: ", var.dtype)
```

```python
var=np.array(["a","b","c","d","T"])
print("Data types: ", var.dtype)
```

```python
var=np.array([1,2,3,4,13,13,14,15,"A","B","C"])
print("Data types: ", var.dtype)
```

```python
x=np.array([1,2,3,4],dtype=np.int8)
print("Data types: ", x.dtype)
print(x)
```

```python
x1=np.array([1,2,3,4],dtype="f")
print("Data types: ", x1.dtype)
print(x1)
```

```python
x2=np.array([1,2,3,4])
new=np.float32(x2)
print("Data types: ", x2.dtype)
print("Data types: ", new.dtype)
print(x2)
print(new)
```

```python
x2=np.array([1,2,3,4])
new=np.float32(x2)
new_one=np.int_(new)
print("Data types: ", x2.dtype)
print("Data types: ", new.dtype)
print("Data types: ", new_one.dtype)
print(x2)
print(new)
print(new_one)
```

```python
x3=np.array([1,2,3,4])
new_1=x3.astype(float)
print(x3)
print(new_1)
```
