# **Join Array**

```python
import numpy as np

var1=np.array([1,2,3,4])
var2=np.array([9,8,7,6])

arr=np.concatenate((var1,var2))
print(arr)

```

```python
var1=np.array([[1,2],[3,4]])
var2=np.array([[9,8],[7,6]])
print(var1)
print()
print(var2)
print()
arr=np.concatenate((var1,var2),axis=1)
print(arr)
```

```python
var_1=np.array([1,2,3,4])
var_2=np.array([9,8,7,6])

arr_new=np.stack((var_1,var_2),axis=0)
arr_new1=np.hstack((var_1,var_2))   #Row
arr_new2=np.vstack((var_1,var_2))   #Column
arr_new3=np.dstack((var_1,var_2))   #Height
print(arr_new)
print()
print(arr_new1)
print()
print(arr_new2)
print()
print(arr_new3)
```

# **Split Array**

```python
var=np.array([1,2,3,4,5,6])
print(var)
ar=np.array_split(var,3)
print()
print(ar)
print(type(ar))
print(ar[0])
```

```python
var1=np.array([[1,2],[3,4],[5,6]])
print(var1)
ar1=np.array_split(var1,3)
ar2=np.array_split(var1,3,axis=1)
print()
print(ar1)
print()
print(ar2)

```
